```python
#Imports
%matplotlib inline
import matplotlib.pyplot as p
import numpy as np

#Constants
dt = 0.08
maxTime = 4
time = np.arange(0,maxTime,dt);#splits the interval of time from 0 to maxT into segments of interval dt
capacitance = 1; 
resistance = 2
threshold = 3
spikeDisplay = 10
tau = 1.0/(resistance*capacitance)
voltage = [0]
injCurrent = (1.0,3.0)
curMag = 4.0

#Functions
def nextVal(ov,dt,dv):
    return(ov+dv*dt)

def dvdt(tau,res,V,I):
    return(((res*I)-V)*(1/tau))

#Plot construction with loop
for j in time: 
    if (injCurrent[0] < j <= injCurrent[1]):
        curCur = curMag
    else:
        curCur = 0
        
    tempV = nextVal(voltage[-1],dt,dvdt(tau,resistance,voltage[-1],curCur)); # creates function for "charging up"
    
    if voltage[-1] == spikeDisplay: 
        voltage.append(0) #when voltage equals the definined heighest point of spike the next value for voltage will be 0
    elif tempV > threshold:
        voltage.append(spikeDisplay) # if our temp voltage is above threshold the next point will be our predefined max
    else: 
        voltage.append(tempV) # if no other conditions are met we will use the tempV as our previous point and solve for our next point with the NextVal function
    
        
p.plot(time,voltage[0:-1])
```
